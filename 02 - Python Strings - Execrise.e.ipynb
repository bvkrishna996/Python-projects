```python
# CODE HERE
# Split text into words
words = var.split()

# Initialize an empty list to store matching words
filtered_words = []

# Loop through each word in the list
for word in words:
    # Convert word to lowercase and check if it starts with 'u' or 'o'
    if word.lower().startswith('u') :
        filtered_words.append(word)

# Print result
print("List of words staring with u is :", filtered_words)
```
