**`Question:` database = {'ram':1234,'raghu':5678,'kanav':1357}**

### 3. show the updated database list

```python
database = {'ram': 1234, 'raghu': 5678, 'kanav': 1357}

# Taking user input
username = input("Enter your username: ")
password = int(input("Enter your password: "))
if username in database and database[username] == password:
    print(f"Hello {username}!")
else:
    # If username and password do not match
    choice = input("Username or password incorrect. Do you want me to add you to the database? (yes/no): ").lower()
    if choice == 'yes':
        new_password = int(input("Enter a new password: "))
        database[username] = new_password
        print(f"User '{username}' added successfully!")
print("Updated Database:", database)
```
