> `Question:` Suppose the cover price of a book is Rs.24.95, but bookstores get a 40% discount. Shipping costs Rs.3 for the first copy and 75 paise for each additional copy. What is the total wholesale cost for 60 copies?

```python
# CODE HERE
cover_price = 24.95
discount = 0.4  # 40% discount
shipping_first_copy = 3
shipping_additional_copy = 0.75
num_copies = 60

# Discount
discounted_price = cover_price * (1 - discount)

# Total books
total_book_cost = num_copies * discounted_price

# Total shipping cost
total_shipping_cost = shipping_first_copy + (num_copies - 1) * shipping_additional_copy

# Total wholesale cost
total_wholesale_cost = total_book_cost + total_shipping_cost

print(f"Total wholesale cost for {num_copies} copies: Rs. {total_wholesale_cost:.2f}")
```
